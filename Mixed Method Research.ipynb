```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.feature_extraction.text import CountVectorizer, ENGLISH_STOP_WORDS
import re

# Set aesthetic style for plots
sns.set_theme(style="whitegrid", palette="muted")
np.random.seed(42)

print("="*50)
print("MIXED METHODS RESEARCH SIMULATION: DATA ANALYSIS")
print("="*50)

# ==========================================
# PHASE 1: QUANTITATIVE ANALYSIS (SURVEY)
# ==========================================
print("\n--- PHASE 1: QUANTITATIVE SURVEY ANALYSIS ---")

# 1. Generate Mock Survey Data (30 respondents, 8 Likert-scale questions 1-5)
n_respondents = 30
survey_df = pd.DataFrame({
    'Q1_Daily_Use': np.random.choice([3, 4, 5], n_respondents, p=[0.2, 0.3, 0.5]),
    'Q2_Increased_Speed': np.random.choice([3, 4, 5], n_respondents, p=[0.1, 0.3, 0.6]),
    'Q3_Improved_Quality': np.random.choice([2, 3, 4], n_respondents, p=[0.2, 0.5, 0.3]),
    'Q4_Reduced_Creativity': np.random.choice([2, 3, 4, 5], n_respondents, p=[0.2, 0.3, 0.3, 0.2]),
    'Q5_Job_Satisfaction': np.random.choice([3, 4, 5], n_respondents, p=[0.2, 0.5, 0.3]),
    'Q6_Employability_Fear': np.random.choice([2, 3, 4], n_respondents, p=[0.3, 0.4, 0.3]),
    'Q7_Trust_in_AI': np.random.choice([1, 2, 3], n_respondents, p=[0.4, 0.4, 0.2]),
    'Q8_Anxiety_Without_AI': np.random.choice([2, 3, 4], n_respondents, p=[0.3, 0.4, 0.3])
})

# 2. Calculate Descriptive Statistics
quant_stats = survey_df.mean().round(2)
quant_std = survey_df.std().round(2)
stats_summary = pd.DataFrame({'Mean': quant_stats, 'Std Dev': quant_std})
print("\nDescriptive Statistics (Survey Means):")
print(stats_summary)

# 3. Inferential Statistics (Correlation between Speed and Satisfaction)
corr_speed_sat, p_val = stats.pearsonr(survey_df['Q2_Increased_Speed'], survey_df['Q5_Job_Satisfaction'])
print(f"\nInferential Stat: Correlation between Speed (Q2) and Satisfaction (Q5)")
print(f"Pearson r = {corr_speed_sat:.3f}, p-value = {p_val:.4f}")

# 4. Quantitative Visualization
fig, ax = plt.subplots(figsize=(10, 6))
means = survey_df.mean().sort_values(ascending=False)
sns.barplot(x=means.values, y=means.index, ax=ax, palette="viridis")
ax.set_title("Quantitative Findings: Developer Perceptions of AI Assistants", fontsize=14)
ax.set_xlabel("Mean Agreement Score (1=Strongly Disagree, 5=Strongly Agree)")
ax.set_xlim(0, 5)
plt.tight_layout()
plt.show()


# ==========================================
# PHASE 2: QUALITATIVE ANALYSIS (INTERVIEWS)
# ==========================================
print("\n--- PHASE 2: QUALITATIVE TEXT ANALYSIS (CODING) ---")

# 1. Mock Interview Transcripts
interviews = [
    "I used to spend time thinking through algorithms. Now I just type a prompt. I feel like a reviewer. My creative muscles are atrophying. I am just stitching AI blocks together. It feels less mine.",
    "It has been a lifeline for me as a junior dev. It helps me understand patterns. I don't feel less creative, I feel augmented. It is like having a senior dev looking over my shoulder.",
    "Productivity is definitely up but I worry about deskilling. Who am I if the AI writes the code? There is a total loss of authorship and professional identity.",
    "Speed is great but debugging AI hallucinations takes so much time. My trust is low. I have to review everything line by line. It creates a lot of anxiety."
]

# 2. Simulate "Open Coding" using NLP (Keyword Extraction)
# We use scikit-learn to find the most frequent meaningful words (codes)
custom_stop_words = list(ENGLISH_STOP_WORDS) + ['feel', 'like', 'just', 'it', 'i', 'my', 'me', 'am']
vectorizer = CountVectorizer(stop_words=custom_stop_words, max_features=10)

# Fit and transform the text data
X = vectorizer.fit_transform(interviews)
words = vectorizer.get_feature_names_out()
word_counts = X.sum(axis=0).A1

# Create a dataframe of "Codes"
codes_df = pd.DataFrame({'Code (Keyword)': words, 'Frequency': word_counts}).sort_values(by='Frequency', ascending=False)
print("\nGenerated Initial Codes (Top Keywords from Transcripts):")
print(codes_df)

# 3. Qualitative Visualization (Code Frequency)
fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x='Frequency', y='Code (Keyword)', data=codes_df, ax=ax, palette="magma")
ax.set_title("Qualitative Findings: Frequency of Initial Codes", fontsize=14)
plt.tight_layout()
plt.show()


# ==========================================
# PHASE 3: MIXED METHODS INTEGRATION
# ==========================================
print("\n--- PHASE 3: DATA INTEGRATION (JOINT DISPLAY) ---")

# Create a Joint Display Table (Crucial for Mixed Methods papers)
# This maps the Quantitative statistical findings to the Qualitative themes
joint_display_data = {
    'Quantitative Finding (Survey)': [
        'High perceived speed (Mean: 4.6)',
        'Moderate creativity loss (Mean: 3.3)',
        'Low trust in AI output (Mean: 1.9)'
    ],
    'Qualitative Theme (Interviews)': [
        'Theme: Cognitive Offloading',
        'Theme: Professional Identity Erosion',
        'Theme: The Reviewer Burden'
    ],
    'Illustrative Qualitative Quote': [
        '"Productivity is definitely up... but I worry about deskilling."',
        '"I am just stitching AI blocks together. It feels less mine."',
        '"Debugging AI hallucinations takes time... I have to review everything."'
    ],
    'Meta-Inference (Integrated Conclusion)': [
        'AI accelerates output but shifts the developer role from creator to manager.',
        'Speed comes at the psychological cost of authorship and creative ownership.',
        'Low trust necessitates heavy manual review, creating hidden time-costs and anxiety.'
    ]
}

joint_display_df = pd.DataFrame(joint_display_data)

# Display the Joint Display in the console
print("\nJoint Display: Integrating Quant and Qual Data")
print(joint_display_df.to_string(index=False))

# Visualize the Joint Display as a table plot for the presentation
fig, ax = plt.subplots(figsize=(12, 6))
ax.axis('tight')
ax.axis('off')
table = ax.table(cellText=joint_display_df.values, colLabels=joint_display_df.columns, cellLoc='center', loc='center')
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1.2, 2.5) # Scale table for readability
plt.title("Joint Display: Integration of Quantitative and Qualitative Findings", fontsize=14, pad=20)
plt.tight_layout()
plt.show()

print("\nAnalysis Complete. Save the generated plots for your presentation and Moodle report.")
```
