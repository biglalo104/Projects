```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#1.Descriptive Statisticss
np.random.seed(42)
sales_data =np.random.normal(loc=500,scale=100,size=200)
sales_data = np.append(sales_data,[500]*20)
print(f"Mean:{np.mean(sales_data):.2f},Std Dev:{np.std(sales_data,ddof=1):.2f}")
```

```python
#2.Inferential Statistics(t_test)
tutored = np.random.normal(loc=85,scale=10,size=30)
non_tutored = np.random.normal(loc=75,scale=12,size=30)
t_stat,p_val_t=stats.ttest_ind(tutored,non_tutored)
print(f"T-test P-value:{p_val_t:.5f}")
```

```python
#3&4.Correlation & Linear Regression
ad_spend = np.random.uniform(1000,5000,50)
sales_rev = ad_spend*5.85 + np.random.normal(0,500,50)
corr=stats.pearsonr(ad_spend,sales_rev)
print(f"Correlation r:{corr.statistic:.2f}")
X_ad=sm.add_constant(ad_spend)
model_lr = sm.OLS(sales_rev,X_ad).fit()
print(f"Linear R-squared:{model_lr.rsquared:.2f}")
```

```python
#6.ANOVA
method_A = np.random.normal(70,5,30)
method_B = np.random.normal(75,6,30)
method_C = np.random.normal(80,5,30)
f_val,p_val_anova=stats.f_oneway(method_A,method_B,method_C)
print(f"ANOVA P-value:{p_val_anova:5f}")
```
